## Page: 📓 Cloud Folders for Staging and Storing Files

> **Course:** 🎯 Create a Database for Zena's Athleisure Idea: create ZENAS_ATHLEISURE_DB owned by SYSADMIN; drop the PUBLIC schema; create a schema called PRODUCTS (also owned by SYSADMIN).

```
%%sql -r create_zenas_db
-- Derived solution
use role {{SYSADMIN_ROLE}};
create database ZENAS_ATHLEISURE_DB;
drop schema ZENAS_ATHLEISURE_DB.PUBLIC;
create schema ZENAS_ATHLEISURE_DB.PRODUCTS;
use warehouse {{WAREHOUSE}};
use schema ZENAS_ATHLEISURE_DB.PRODUCTS;
```

> **Course:** 🥋 Create an Internal Stage and Load the Sweatsuit Files Into It: download and unzip sweatsuits.zip. UI steps (DLKW_003-008): Catalog → ZENAS_ATHLEISURE_DB → PRODUCTS → Create → Stage → **Snowflake Managed**; Stage Name **sweatsuits**; Directory table **on**; Encryption **Server-side**; Create. Then **+ Files** and add the 10 unzipped PNG files.

```
%%sql -r create_sweatsuits_stage
-- SQL equivalent of the Create Stage dialog (DLKW_005): directory table on, server-side encryption
create stage ZENAS_ATHLEISURE_DB.PRODUCTS.SWEATSUITS
  directory = (enable = true)
  encryption = (type = 'SNOWFLAKE_SSE');
```

```
%%sql -r load_sweatsuits
-- SQL equivalent of **+ Files**: the unzipped files sit on the Downloads stand-in stage
copy files into @ZENAS_ATHLEISURE_DB.PRODUCTS.SWEATSUITS
from @UTIL_DB.PUBLIC.BADGE4_DOWNLOADS
files = ('burgundy_sweatsuit.png', 'charcoal_grey_sweatsuit.png', 'forest_green_sweatsuit.png', 'navy_blue_sweatsuit.png', 'orange_sweatsuit.png', 'pink_sweatsuit.png', 'purple_sweatsuit.png', 'red_sweatsuit.png', 'royal_blue_sweatsuit.png', 'yellow_sweatsuit.png');
alter stage ZENAS_ATHLEISURE_DB.PRODUCTS.SWEATSUITS refresh;
list @ZENAS_ATHLEISURE_DB.PRODUCTS.SWEATSUITS;
```

## Page: 🎯 Create Another Stage on Your Own

> **Course:** 🎯 Create Another Internal Stage: in the PRODUCTS schema, owned by SYSADMIN, named PRODUCT_METADATA, **client-side encryption**. Load the files from metadata.zip (DLKW_002: 3 files). At the end you should have one database, one dropped and one created schema (2 schemas), and 2 stages.

```
%%sql -r create_product_metadata_stage
-- SQL equivalent of the Create Stage dialog (Snowflake Managed, client-side encryption). DLKW_009's icons show no directory table on this stage.
create stage ZENAS_ATHLEISURE_DB.PRODUCTS.PRODUCT_METADATA
  encryption = (type = 'SNOWFLAKE_FULL');
```

```
%%sql -r load_product_metadata
-- SQL equivalent of **+ Files**
copy files into @ZENAS_ATHLEISURE_DB.PRODUCTS.PRODUCT_METADATA
from @UTIL_DB.PUBLIC.BADGE4_DOWNLOADS
files = ('product_coordination_suggestions.txt', 'sweatsuit_sizes.txt', 'swt_product_line.txt');
list @ZENAS_ATHLEISURE_DB.PRODUCTS.PRODUCT_METADATA;
```
