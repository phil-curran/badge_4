## Page: 🎯 Camilla's GeoSpatial Data (Part 1)

> **Course:** 🎯 Put Your Snowflake Skills to Work! Everything owned by SYSADMIN: create MELS_SMOOTHIE_CHALLENGE_DB; drop the PUBLIC schema; add schema TRAILS; add internal named stages TRAILS_GEOJSON and TRAILS_PARQUET (client-side encryption is fine) in the TRAILS schema (DLKW_075).

```
%%sql -r create_mels_db
-- Derived solution
use role {{SYSADMIN_ROLE}};
create database MELS_SMOOTHIE_CHALLENGE_DB;
drop schema MELS_SMOOTHIE_CHALLENGE_DB.PUBLIC;
create schema MELS_SMOOTHIE_CHALLENGE_DB.TRAILS;
use warehouse {{WAREHOUSE}};
use schema MELS_SMOOTHIE_CHALLENGE_DB.TRAILS;
```

```
%%sql -r create_trails_stages
-- Derived solution. DLKW_075's icons show both stages with a directory table.
create stage TRAILS_GEOJSON
  directory = (enable = true)
  encryption = (type = 'SNOWFLAKE_FULL');
create stage TRAILS_PARQUET
  directory = (enable = true)
  encryption = (type = 'SNOWFLAKE_FULL');
```
