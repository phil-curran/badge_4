# Badge 4: DORA checks

Every DORA step from the course, in order. Each check has two cells:

1. **Local check**: the course's check query with the external `GRADER` call removed. It runs entirely in this account and sends nothing.
2. **DORA (exact course code)**: verbatim and unedited (DORA audits a hash of the code).

⚠️ **Running a DORA cell sends data outside this account** to Snowflake Education's grading service. Run only the local checks unless you intend to submit.

Each check is meant to be run at the end of its lesson: later lessons add stages and views that earlier checks count (for example DLKW05 counts the stages and file formats in Mel's database).

## DLKW01 (after `lesson_02_data_structures_and_stages`)

```
%%sql -r dlkw01_local
-- Local check: same query without the GRADER call
select step, actual, expected, (actual = expected) as passed, description from
(
 SELECT
 'DLKW01' as step
  ,( select count(*)  
      from ZENAS_ATHLEISURE_DB.INFORMATION_SCHEMA.STAGES 
      where stage_schema = 'PRODUCTS'
      and 
      (stage_type = 'Internal Named' 
      and stage_name = ('PRODUCT_METADATA'))
      or stage_name = ('SWEATSUITS')
   ) as actual
, 2 as expected
, 'Zena stages look good' as description
); 
```

```
%%sql -r dlkw01_dora
select GRADER(step, (actual = expected), actual, expected, description) as graded_results from
(
 SELECT
 'DLKW01' as step
  ,( select count(*)  
      from ZENAS_ATHLEISURE_DB.INFORMATION_SCHEMA.STAGES 
      where stage_schema = 'PRODUCTS'
      and 
      (stage_type = 'Internal Named' 
      and stage_name = ('PRODUCT_METADATA'))
      or stage_name = ('SWEATSUITS')
   ) as actual
, 2 as expected
, 'Zena stages look good' as description
); 
```

## DLKW02 (after `lesson_03_leaving_data_where_it_lands`)

```
%%sql -r dlkw02_local
-- Local check: same query without the GRADER call
select step, actual, expected, (actual = expected) as passed, description from
(
 SELECT
   'DLKW02' as step
   ,( select sum(tally) from
        ( select count(*) as tally
        from ZENAS_ATHLEISURE_DB.PRODUCTS.SWEATBAND_PRODUCT_LINE
        where length(product_code) > 7 
        union
        select count(*) as tally
        from ZENAS_ATHLEISURE_DB.PRODUCTS.SWEATSUIT_SIZES
        where LEFT(sizes_available,2) = char(13)||char(10))     
     ) as actual
   ,0 as expected
   ,'Leave data where it lands.' as description
); 
```

```
%%sql -r dlkw02_dora
select GRADER(step, (actual = expected), actual, expected, description) as graded_results from
(
 SELECT
   'DLKW02' as step
   ,( select sum(tally) from
        ( select count(*) as tally
        from ZENAS_ATHLEISURE_DB.PRODUCTS.SWEATBAND_PRODUCT_LINE
        where length(product_code) > 7 
        union
        select count(*) as tally
        from ZENAS_ATHLEISURE_DB.PRODUCTS.SWEATSUIT_SIZES
        where LEFT(sizes_available,2) = char(13)||char(10))     
     ) as actual
   ,0 as expected
   ,'Leave data where it lands.' as description
); 
```
